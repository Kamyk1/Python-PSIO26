W Pythonie, korzystając z pętli (ver.1) i rozszerzonego operatora zakresu (ver.2), przygotuj formuły dla:

1. lustrzanego odbicia obrazu w poziomie
2. lustrzanego odbicia obrazu w pionie
3. obrotu obrazu o 90° w prawo oraz 90° w lewo
4. obrót obrazu o 180°
5. rozszerzenia macierzy do kwadratu MxM (jeżeli M>N) - kolumny po lewej i prawej stronie są wyzerowane
6. wycięcia z macierzy kwadratu NxN (jeżeli M>N) - wycinamy środkową część macierzy

```python
import numpy as np
#conda update -n base -c defaults conda

obraz = np.array([[1,2,3,16],[4,5,6,17],[7,8,9,19],[10,11,12,20]
])
```

```python
# 1. Lustrzanego odbicia obrazu w poziomie.

def odbicie_obrazu_w_poziomie(obraz_we):
    return obraz_we[::-1,:]

print(odbicie_obrazu_w_poziomie(obraz))
```

```python
# 2. Lustrzanego odbicia obrazu w pionie.

def odbicie_obrazu_w_pionie(obraz_we):
    return obraz_we[:,::-1]

print(odbicie_obrazu_w_pionie(obraz))
```

```python
# 3. Obrotu obrazu o 90° w prawo oraz 90° w lewo.
def Obrot_obrazu_90_prawo(obraz_we):
    return obraz_we.T[:,::-1]

print("Obrot w lewo:\n",Obrot_obrazu_90_prawo(obraz))

def Obrot_obrazu_90_lewo(obraz_we):
    return obraz_we.T[::-1,:]

print("Obrot w prawo:\n",Obrot_obrazu_90_lewo(obraz))
```

```python
# 4. Obrót obrazu o 180°.
def Obrot_obrazu_180(obraz_we):
    return obraz_we[::-1,::-1]

print("Obrot o 180\n",Obrot_obrazu_180(obraz))
```

```python
# 5. rozszerzenia macierzy do kwadratu MxM (jeżeli M>N) - kolumny po lewej i prawej stronie są wyzerowane.
def rozszerzenie_macierzy(obraz_we):
    H = len(obraz_we[:,1])
    W = len(obraz_we[1,:])
    if W < H:
        wynik = np.zeros([H,H])
        indeks = (H - W) // 2
        wynik[:,indeks:indeks+W] = obraz_we
        return wynik
    return obraz_we

print("Rozszerzenie Macierzy:\n",rozszerzenie_macierzy(obraz))
```

```python
# 6. wycięcia z macierzy kwadratu NxN (jeżeli M>N) - wycinamy środkową część macierzy.
def wyciecie_macierzy(obraz_we):
    H = len(obraz_we[:,1])
    W = len(obraz_we[1,:])
    if H == W:
        return obraz_we

    if H > W:
        indeks = (H - W) // 2
        return obraz_we[indeks:indeks+W,:]

    if W > H:
        indeks = (W-H) // 2
        return obraz_we[:,indeks:indeks+H]

print("Wyciecie macierzy:\n",wyciecie_macierzy(obraz))
```
